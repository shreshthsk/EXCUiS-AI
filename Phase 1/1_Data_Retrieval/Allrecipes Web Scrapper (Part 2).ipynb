```python
import json

# Load both datasets
with open('recipes.json', 'r') as f:
    instructions_data = json.load(f)

with open('recipes_ingredients.json', 'r') as f:
    ingredients_data = json.load(f)

# Ensure both are lists (if they are a list of recipes)
if isinstance(instructions_data, dict):
    instructions_data = [instructions_data]
if isinstance(ingredients_data, dict):
    ingredients_data = [ingredients_data]

# Create a merged dictionary using title as the key
merged_recipes = {}

# First, add from ingredients dataset
for entry in ingredients_data:
    title = entry.get("title", "").strip()
    merged_recipes[title] = {
        "title": title,
        "ingredients": entry.get("ingredients", []),
        "instructions": []  # to be filled later if available
    }

# Then, update or add from instructions dataset
for entry in instructions_data:
    title = entry.get("title", "").strip()
    if title in merged_recipes:
        merged_recipes[title]["instructions"] = entry.get("instructions", [])
    else:
        merged_recipes[title] = {
            "title": title,
            "ingredients": [],
            "instructions": entry.get("instructions", [])
        }

# Convert back to a list
merged_list = list(merged_recipes.values())

# Save the merged dataset
with open('merged_recipes.json', 'w') as f:
    json.dump(merged_list, f, indent=2)

print("✅ Merged recipe data saved to 'merged_recipes.json'")
```
