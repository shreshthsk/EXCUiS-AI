```python
import pandas as pd

# Load your final merged dataset
df = pd.read_json("merged_recipes.json")

# 1. Check for missing values
missing_summary = df.isnull().sum()

# 2. Check for empty strings or very short lists
empty_ingredients = df[df['ingredients'].apply(lambda x: len(x) < 3)]
# Make sure 'recipe' is treated as a string before stripping
empty_recipes = df[df['instructions'].apply(lambda x: isinstance(x, str) and len(x.strip()) < 30)]

# 3. Normalize for easier spotting
df['name_clean'] = df['title'].str.strip().str.lower()
duplicates = df[df.duplicated('name_clean', keep=False)]

# Show results
print("Missing Values:\n", missing_summary)
print("\nEntries with few ingredients:\n", empty_ingredients[['title', 'ingredients']])
print("\nEntries with short recipes:\n", empty_recipes[['title', 'instructions']])
print("\nPossible Duplicates:\n", duplicates[['title']])
```

```python
df_cleaned = df[df['ingredients'].apply(lambda x: isinstance(x, list) and len(x) >= 3)]
```

```python
df_dropped = df[df['ingredients'].apply(lambda x: not isinstance(x, list) or len(x) < 3)]
df_dropped.to_csv("dropped_due_to_ingredients.csv", index=False)
```
