##ANOVA (Analysis of Variance) checks:

Null Hypothesis (H₀): The mean flavor scores based on the ingredients used are the same across all regions.

Alternative Hypothesis (H₁): The mean flavor scores based on the ingredients used differ for at least one region.

**Note: This analysis captures regional trends in the usage of ingredients associated with different flavor profiles. It does not directly measure the final taste intensity of dishes, as ingredient quantities and preparation methods were not accounted for.
For example, a higher occurrence of bitter ingredients (like fenugreek seeds) in Gujarati recipes suggests a regional preference or inclusion of such flavors, but does not quantify the bitterness of the final dish.

Result given by ANOVA:

F-statistic: How much the groups differ relative to the variability inside the groups.

p-value: Probability that differences happened by random chance.

If p-value < 0.05 → We reject the null hypothesis (meaning there is a significant difference).

If p-value > 0.05 → We fail to reject the null hypothesis (meaning there is no significant difference).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('mapped_dishes_with_flavors_and_region.csv')

# Display basic info about the dataset
print("Basic Information about the dataset:")
print(df.info())

# Check for missing data
print("\nMissing Data:")
print(df.isnull().sum())

# Descriptive statistics of the flavor columns
print("\nDescriptive Statistics of Flavor Profiles:")
print(df[['salty', 'spicy', 'sweet', 'sour', 'bitter']].describe())

# Regional trend analysis: Mean flavor profiles per region
region_flavor_means = df.groupby('region')[['salty', 'spicy', 'sweet', 'sour', 'bitter']].mean()
```

```python
import scipy.stats as stats

# Perform one-way ANOVA for each flavor profile
flavors = ['salty', 'spicy', 'sweet', 'sour', 'bitter','umami']

for flavor in flavors:
    f_stat, p_val = stats.f_oneway(*[df[df['region'] == region][flavor] for region in df['region'].unique()])
    print(f"{flavor} - F-statistic: {f_stat}, p-value: {p_val}")
```

##📄 Conclusion:

The p-values for all flavors are greater than 0.05, so we fail to reject the null hypothesis for each flavor. This indicates that, statistically, there is no significant difference in the usage of any particular flavor across the regions based on our dataset.
While this result might seem counterintuitive, it emphasizes the complexity and nuance of Indian cuisine—regional identity may not solely lie in flavor profiles but in ingredient quantities, preparation styles, and combinations.
