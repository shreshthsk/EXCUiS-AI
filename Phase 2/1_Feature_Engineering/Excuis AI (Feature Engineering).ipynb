```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import json

# Load JSON data properly - handle cases where file might contain multiple JSON objects
with open("cleaned_recipes.json", 'r') as f:
    # This handles both single JSON object and multiple JSON objects (JSON lines format)
    data = [json.loads(line) for line in f]

# Convert to DataFrame
df = pd.DataFrame(data)

# Alternative if the file is a proper JSON array:
# df = pd.read_json("cleaned_recipes.json", orient='records')

# Step 1: Handle unknowns
df['region'] = df['region'].fillna('unknown')

# Step 2: Encoder with UNKNOWN support
label_encoder = LabelEncoder()
known_regions = df[df['region'] != 'unknown']['region'].unique()
label_encoder.fit(known_regions)

# Step 3: Transform with fallback for unknowns
df['region_code'] = df['region'].apply(
    lambda x: -1 if x == 'unknown' else label_encoder.transform([x])[0]
)

# Step 4: Save the mapping
region_mapping = {name: code for code, name in enumerate(label_encoder.classes_)}
region_mapping['unknown'] = -1

print("Encoded Data:")
print(df[['region', 'region_code']].head())  # Show only first few rows
print("\nRegion Mapping:", region_mapping)
```

```python
df.to_json("Excuis_AI.json")
```
